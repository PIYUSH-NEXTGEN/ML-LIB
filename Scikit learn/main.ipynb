```python
# Import the built-in Breast Cancer dataset from scikit-learn
from sklearn.datasets import load_breast_cancer

# Import the function used to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import StandardScaler to standardize/scale the feature values
from sklearn.preprocessing import StandardScaler

# Import KNeighborsClassifier, the K-Nearest Neighbors classification algorithm
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Load the Breast Cancer dataset and directly return its features (X) and target labels (y)
X, y = load_breast_cancer(return_X_y=True)
```

```python
# Split X and y into training data (80%) and testing data (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
# Create a StandardScaler object that will be used to standardize the features
scaler = StandardScaler()

# Learn the mean and standard deviation from the training data, then scale the training data
X_train_scaled = scaler.fit_transform(X_train)

# Scale the test data using the mean and standard deviation learned from the training data
X_test_scaled = scaler.transform(X_test)
```

```python
# Create a K-Nearest Neighbors classifier using its default parameters
knn = KNeighborsClassifier()

# Train the KNN classifier using the scaled training features and their corresponding labels
knn.fit(X_train_scaled, y_train)
```

```python
print(knn.score(X_test_scaled, y_test))
```

```python
# Import the built-in Breast Cancer dataset from scikit-learn
from sklearn.datasets import load_breast_cancer

# Import the function used to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import StandardScaler for standardizing numerical features
from sklearn.preprocessing import StandardScaler


# Load the Breast Cancer dataset
# return_X_y=True tells sklearn to directly return:
# X = features
# y = target labels
X, y = load_breast_cancer(return_X_y=True)


# Split the data into training and testing sets
# X_train → training features
# X_test  → testing features
# y_train → labels for training data
# y_test  → labels for testing data
# test_size=0.2 → 20% test data and 80% training data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2
)


# Create a StandardScaler object
# This object will learn the mean and standard deviation
# of each feature from the training data
scaler = StandardScaler()


# fit_transform() does two things:
#
# 1. fit()       → learn mean and standard deviation from X_train
# 2. transform() → standardize X_train using those learned values
#
# The result is stored in X_train_scaled
X_train_scaled = scaler.fit_transform(X_train)


# Transform the test data using the SAME mean and standard deviation
# learned from X_train.
#
# We use transform(), NOT fit_transform(), because fitting on the
# test data would cause data leakage.
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Create the KNN classifier
clf = KNeighborsClassifier()

# Train the classifier using the scaled training data
clf.fit(X_train_scaled, y_train)
```

```python
# Evaluate the classifier on the scaled test data
# For KNeighborsClassifier, this returns accuracy
clf.score(X_test_scaled, y_test)
```
